Dünyanın en çok ziyaret edilen şehirlerinden biri olan New York City’e hoş geldin. New York’ta, seyahat edenlerin birkaç geceden birkaç aya kadar değişebilen kısa süreli konaklama ihtiyaçlarını karşılamak için çok sayıda Airbnb ilanı var. Bu projede, `.csv`, `.tsv` ve `.xlsx` gibi birden fazla dosya türünden veri birleştirerek New York’taki Airbnb pazarına daha yakından bakacağız.

Unutma, **CSV**, **TSV** ve **Excel** dosyaları verileri saklamak için yaygın biçimlerdir.
2019 Airbnb ilanlarına ait verileri içeren üç dosya elinde:

**data/airbnb_price.csv**
Bu, Airbnb ilan fiyatları ve konumlarına dair veriler içeren bir CSV dosyasıdır.
- **`listing_id`**: ilan için benzersiz tanımlayıcı
- **`price`**: gecelik ilan fiyatı (USD)
- **`nbhood_full`**: ilanın bulunduğu ilçe ve mahalle adı

**data/airbnb_room_type.xlsx**
Bu, Airbnb ilan açıklamaları ve oda türlerine dair veriler içeren bir Excel dosyasıdır.
- **`listing_id`**: ilan için benzersiz tanımlayıcı
- **`description`**: ilan açıklaması
- **`room_type`**: Airbnb’de üç tür oda vardır: shared rooms, private rooms ve entire homes/apartments

**data/airbnb_last_review.tsv**
Bu, Airbnb ev sahibi adları ve son değerlendirme tarihlerine dair veriler içeren bir TSV dosyasıdır.
- **`listing_id`**: ilan için benzersiz tanımlayıcı
- **`host_name`**: ilanın ev sahibinin adı
- **`last_review`**: ilanın en son değerlendirildiği tarih


```python
import pandas as pd
import numpy as np

# 1. Dosyayı doğrudan okuyoruz (DataCamp'te genellikle dosya 'data/' klasöründedir)
df_review = pd.read_csv('data/airbnb_last_review.tsv', sep='\t')

# 2. Python'a bunun bir tarih sütunu olduğunu söylüyoruz
df_review['last_review'] = pd.to_datetime(df_review['last_review'])

# 3. Senden istenen o iki ayrı değişkene tercih ettiğin isimleri veriyorsun
en_erken_degerlendirme = df_review['last_review'].min()
en_yakin_degerlendirme = df_review['last_review'].max()

print("En erken değerlendirme:", en_erken_degerlendirme)
print("En yakın değerlendirme:", en_yakin_degerlendirme)

df_room_type = pd.read_excel('data/airbnb_room_type.xlsx')

# 1. Önce sütundaki tüm yazıları küçük harfe çevirelim (veriyi temizleyelim)
df_room_type["room_type"] = df_room_type["room_type"].str.lower()

# 2. Şimdi sadece "private room" olanları sayalım. 
# .sum() veya len() kullanabiliriz, ama veri filtreleyip uzunluğunu almak çok yaygındır:
nb_private_rooms = len(df_room_type[df_room_type["room_type"] == "private room"])

print("Özel Oda Sayısı:", nb_private_rooms)


# Ortalama ilan fiyatı nedir? En yakın iki ondalığa yuvarlayın ve bir değişkende saklayın.

df_price = pd.read_csv("data/airbnb_price.csv")

# 2. Fiyat sütunundaki metinleri temizle ve sayıya çevir
# Not: Eğer veri setinizde "$" varsa ' dollars' kısmını '$' ile değiştirin.
df_price["price"] = df_price["price"].str.replace(' dollars', '')
df_price["price"] = df_price["price"].astype(float)

# Ortalamayı al ve yuvarla
price = df_price["price"]
avg_price = round(price.mean(), 2)

print(avg_price)


#Yeni değişkenleri, sırasıyla şu dört sütunu içerecek şekilde review_dates adlı tek bir DataFrame’de birleştirin: first_reviewed, last_reviewed, nb_private_rooms ve avg_price. DataFrame yalnızca tek bir satır değer içermelidir.

review_dates = pd.DataFrame({"first_reviewed": [en_erken_degerlendirme],
                            "last_reviewed": [en_yakin_degerlendirme],
                            "nb_private_rooms":[nb_private_rooms],
                            "avg_price":[avg_price]})
print(review_dates)

```
